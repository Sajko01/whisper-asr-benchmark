## Doprinos 2

```python
from faster_whisper import WhisperModel
import psutil
import os
import time
import pandas as pd
```

```python
def ram_mb():
    proces = psutil.Process(os.getpid())
    return proces.memory_info().rss / (1024 * 1024)
```

```python
ram_pre = ram_mb()

model_int8 = WhisperModel(
    "base",
    device="cpu",
    compute_type="int8"
)

ram_posle = ram_mb()
ram_int8 = ram_posle - ram_pre

start = time.perf_counter()

segments, info = model_int8.transcribe(
    "audio/govor60.wav",
    language="en"
)

tekst_int8 = ""
for segment in segments:
    tekst_int8 += segment.text + " "

kraj = time.perf_counter()
vreme_int8 = kraj - start

print("INT8 vreme:", round(vreme_int8, 2), "s")
print("INT8 RAM:", round(ram_int8, 2), "MB")
```

```python
from faster_whisper import WhisperModel
import psutil
import os
import time

def ram_mb():
    proces = psutil.Process(os.getpid())
    return proces.memory_info().rss / (1024 * 1024)

ram_pre = ram_mb()

model_float32 = WhisperModel(
    "base",
    device="cpu",
    compute_type="float32"
)

ram_posle = ram_mb()
ram_float32 = ram_posle - ram_pre

start = time.perf_counter()

segments, info = model_float32.transcribe(
    "audio/govor60.wav",
    language="en"
)

tekst_float32 = ""
for segment in segments:
    tekst_float32 += segment.text + " "

kraj = time.perf_counter()
vreme_float32 = kraj - start

print("FLOAT32 vreme:", round(vreme_float32, 2), "s")
print("FLOAT32 RAM:", round(ram_float32, 2), "MB")
```

```python
import pandas as pd

tabela_doprinos2 = pd.DataFrame({
    "Quantization": ["float32", "int8"],
    "Vreme [s]": [
        8.23,
        5.02
    ],
    "RAM [MB]": [
        312.81,
        113.80
    ]
})

tabela_doprinos2
```

```python
tabela_doprinos2.to_csv(
    "rezultati/doprinos2_kvantizacija.csv",
    index=False,
    encoding="utf-8-sig"
)
```

## Komentar rezultata

Rezultati prikazani u tabeli pokazuju značajan uticaj kvantizacije na performanse Faster-Whisper sistema.

Nekvantizovana `float32` varijanta ostvarila je vreme izvršavanja od 8.23 s uz memorijsku potrošnju od 312.81 MB. Sa druge strane, kvantizovana `int8` varijanta izvršila je transkripciju za svega 5.02 s, uz potrošnju od 113.80 MB RAM memorije.

Poređenjem rezultata može se uočiti da je `int8` režim približno 39% brži od `float32` varijante. Istovremeno, memorijska potrošnja smanjena je za približno 199 MB, odnosno oko 64%.

Dobijeni rezultati potvrđuju da kvantizacija predstavlja veoma efikasan način optimizacije sistema za automatsko prepoznavanje govora. Na korišćenom CPU okruženju `int8` režim omogućava značajno kraće vreme izvršavanja i znatno manju potrošnju memorije, što ga čini pogodnijim za praktične primene na računarima sa ograničenim hardverskim resursima.
